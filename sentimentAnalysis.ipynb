```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, classification_report
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import re
from nltk.tokenize import word_tokenize
```

```python
df=pd.read_csv('sentiment_analysis.csv')
df
```

```python
text= df[['text', 'sentiment']]
```

```python
text
```

```python
data=text.sample(frac=1).reset_index(drop=True)
data
```

```python
x=data['text']
y=data['sentiment']
```

```python
nltk.download('stopwords')
en_english=stopwords.words('english')
```

```python
lowercase= x.str.lower().apply(lambda x: re.sub('[^\w\s]','',x))
lowercase
```

```python
no_stopwords= lowercase.apply(lambda x: ' '.join([ word for word in x.split() if word not in en_english]))
no_stopwords
```

```python
countvec= CountVectorizer()
bow=countvec.fit_transform(no_stopwords).toarray()
bow
```

```python
df=pd.DataFrame(bow, columns=countvec.get_feature_names_out())
df
```

```python
y
```

```python
X_train, X_test, y_train, y_test= train_test_split(df, y, test_size=0.2, random_state=42)
```

```python
model= LogisticRegression()
model.fit(X_train, y_train)
```

```python
prediction= model.predict(X_test)
```

```python
prediction
```

```python
accuracy_score(prediction,y_test)
```

```python
print(classification_report(y_test, prediction,zero_division=0))
```

```python
t= countvec.transform(["i had an accident"])
t.toarray()
```

```python
result=model.predict(t.toarray())
result
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
nb= MultinomialNB().fit(X_train, y_train)
nb_prediction= nb.predict(X_test)
```

```python
accuracy_score(y_test, nb_prediction)
```

```python
print(classification_report(y_test, nb_prediction,zero_division=0))
```
